### Case Folding

Case Folding digunakan untuk mengubah semua huruf menjadi huruf kecil dan menghapus angka, tanda baca, hastag, nomor, emoji, whitespace dan URL.

```python
import re

# Compile emoji
EMOJI_PATTERN = re.compile(
    "["
    "\U0001F600-\U0001F64F"  # emoticon
    "\U0001F300-\U0001F5FF"  # simbol & gambar
    "\U0001F680-\U0001F6FF"  # simbol transportasi & peta
    "\U0001F1E0-\U0001F1FF"  # bendera (iOS)
    "\U00002702-\U000027B0"  # Dingbat
    "\U000024C2-\U0001F251"  # Huruf dalam kurung dan simbol-simbol lainnya
    "]+",
    flags=re.UNICODE
)

def clean_text(text):
    text = text.lower() # Lowercasing (Mengubah huruf besar menjadi huruf kecil)
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE) # Hapus URL
    text = re.sub(r'@\w+', '', text) # Hapus mention
    text = re.sub(r'#\w+', '', text) # Hapus hashtag
    text = EMOJI_PATTERN.sub(r'', text) # Hapus emoji dengan (compiled pattern)
    text = re.sub(r'[\d+\.\!\?\,\;\:"\)\(/\[\]\\-]+', '', text) # Hapus tanda baca dan angka
    text = re.sub(r'\s+', ' ', text).strip() # Hapus spasi kosong yang berlebihan
    return text

df_analisis['text_clean'] = df_analisis['text'].apply(clean_text)
display(df_analisis[['text', 'text_clean']].head())
```

Tahap case folding mengubah seluruh karakter huruf menjadi huruf kecil (*lowercase*) untuk menyeragamkan bentuk kata. Selain itu, dilakukan juga pembersihan teks dengan menghapus URL, mention, hashtag, emoji, angka, dan tanda baca yang tidak diperlukan.

### Tokenizing dan Stopwords Removal

**Tokenizing:** Memecah kalimat menjadi kata-kata tunggal (token).

**Stopword Removal:** Menghapus kata-kata sambung atau kata yang tidak memiliki makna penting dalam analisis

```python
import nltk
from nltk.corpus import stopwords

# Download data NLTK yang diperlukan
try:
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('stopwords', quiet=True)

# Kombinasikan Stopword english dan bahasa indonesia
english_stopwords = set(stopwords.words('english'))
indonesian_stopwords = set(stopwords.words('indonesian'))
custom_stopwords = {
    'yg', 'rp', 'aja', 'ga', 'ya', 'gak', 'nya', 'utk',
    'dgn', 'jg', 'sy', 'krn', 'tdk', 'tp', 'org',
    'udah', 'sdh', 'itu', 'ini', 'ada', 'juga', 'bisa',
    'kalo', 'kalau', 'gimana', 'emang', 'kayak', 'banget',
    'dong', 'sih', 'deh', 'loh', 'kan', 'nih', 'tuh'
}
all_stopwords = english_stopwords.union(indonesian_stopwords).union(custom_stopwords)

def tokenize_and_remove_stopwords(text):
    if pd.isna(text): # Handle nilai NaN jika ada
        return ""
    tokens = text.split() # Tokenisasi sederhana dengan memisahkan berdasarkan spasi
    filtered_tokens = [word for word in tokens if word not in all_stopwords and len(word) > 1] # Hapus kata-kata pengisi dan kata-kata yang terdiri dari satu karakter
    return ' '.join(filtered_tokens)

df_analisis['text_processed'] = df_analisis['text_clean'].apply(tokenize_and_remove_stopwords)
display(df_analisis[['text', 'text_clean', 'text_processed']].head())
```

Di tahap ini dilakukan tokenisasi dengan memecah teks menjadi kumpulan kata berdasarkan spasi. Selanjutnya dilakukan stopword removal menggunakan kombinasi stopword Bahasa Indonesia dan Bahasa Inggris dari NLTK serta stopword tambahan yang disesuaikan dengan karakteristik bahasa informal pada media sosial X (Twitter). Tujuannya menghilangkan kata-kata umum yang tidak memiliki penting terhadap penentuan sentimen sehingga hanya tersisa kata-kata yang lebih representatif untuk proses analisis sentimen.

### Stemming

Stemming merupakan proses pengurangan
imbuhan dalam kata-kata ke bentuk dasarnya
sehingga membantu dalam pre-processing teks, kata,
dan dokumen untuk normalisasi teks.

```python
%%capture
!pip install Sastrawi

import Sastrawi
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# Create stemmer
factory = StemmerFactory()
stemmer = factory.create_stemmer()

def stem_text(text):
    if pd.isna(text):
        return ""
    return stemmer.stem(text)

df_analisis['text_stemmed'] = df_analisis['text_processed'].apply(stem_text)
```

```python
display(df_analisis[['text', 'text_clean', 'text_processed', 'text_stemmed']].head())
```

Proses stemming menggunakan library Sastrawi berhasil merubah kata ke bentuk dasarnya — misalnya "melemahkan", "pelemahan", dan "dilemahkan" seluruhnya menjadi satu token tunggal "lemah". Perubahan ini krusial untuk memastikan bahwa kata-kata yang secara identik tidak dihitung sebagai entitas terpisah dalam analisis frekuensi dan pemodelan topik.
Namun terdapat batasan metodologis: Sastrawi dirancang untuk bahasa Indonesia formal, sehingga kata-kata gaul khas Twitter seperti "anjlok", "gw", atau "bgt" tidak mengalami stemming yang bermakna dan tetap dipertahankan dalam bentuk aslinya — yang justru menjadi keuntungan karena nuansa informal tersebut tetap terjaga untuk keperluan analisis emosi dan sentimen.

```python
from nltk.util import ngrams
from collections import Counter

# Combine all stemmed text into a single string for N-gram analysis
all_stemmed_text_for_ngrams = ' '.join(df_analisis['text_stemmed'].dropna())

# Tokenize the stemmed text
tokenized_stemmed_words = all_stemmed_text_for_ngrams.split()

# Generate bigrams (2-word phrases)
bigrams_stemmed = list(ngrams(tokenized_stemmed_words, 2))
common_bigrams_stemmed = Counter(bigrams_stemmed).most_common(20)

# Convert to DataFrame for visualization
common_bigrams_stemmed_df = pd.DataFrame([(' '.join(word), count) for word, count in common_bigrams_stemmed], columns=['bigram', 'count'])

plt.figure(figsize=(12, 7))
sns.barplot(x='count', y='bigram', data=common_bigrams_stemmed_df)
plt.title('20 Most Common Stemmed Bigrams in Tweets')
plt.xlabel('Count')
plt.ylabel('Bigram')
plt.tight_layout()
plt.show()

# Generate trigrams (3-word phrases)
trigrams_stemmed = list(ngrams(tokenized_stemmed_words, 3))
common_trigrams_stemmed = Counter(trigrams_stemmed).most_common(20)

# Convert to DataFrame for visualization
common_trigrams_stemmed_df = pd.DataFrame([(' '.join(word), count) for word, count in common_trigrams_stemmed], columns=['trigram', 'count'])

plt.figure(figsize=(12, 7))
sns.barplot(x='count', y='trigram', data=common_trigrams_stemmed_df)
plt.title('20 Most Common Stemmed Trigrams in Tweets')
plt.xlabel('Count')
plt.ylabel('Trigram')
plt.tight_layout()
plt.show()
```

### Pre-trained model

```python
from transformers import pipeline
import torch

sentiment_model_loaded = False
try:
    # Load a publicly available pre-trained Indonesian sentiment analysis model
    sentiment_pipeline = pipeline(
        'sentiment-analysis',
        model='w11wo/indonesian-roberta-base-sentiment-classifier',
        tokenizer='w11wo/indonesian-roberta-base-sentiment-classifier',
        device=0 if torch.cuda.is_available() else -1 # Use GPU if available
    )
    sentiment_model_loaded = True
    print("Model sentimen berhasil dimuat: w11wo/indonesian-roberta-base-sentiment-classifier")
except Exception as e:
    print(f"Gagal memuat model sentimen: {e}")
    print("Analisis sentimen akan dilewati karena masalah otorisasi/akses model atau model tidak ditemukan.")

if sentiment_model_loaded:
    # Define a function to get sentiment, handling potential NaN values
    def get_sentiment(text):
        if pd.notna(text):
            # Apply sentiment analysis, truncating long texts to prevent errors
            result = sentiment_pipeline(text, truncation=True, max_length=512)[0]
            return result['label'], result['score']
        return None, None

    # Apply the function to 'text_stemmed' column and create new 'sentimen' and 'skor_sentimen' columns
    # Using .loc to avoid SettingWithCopyWarning if df_analisis is a slice
    df_analisis[['sentimen', 'skor_sentimen']] = df_analisis['text_clean'].apply(lambda x: pd.Series(get_sentiment(x)))

    # Tampilkan distribusi sentimen
    print("\nDistribusi Sentimen:")
    sentiment_counts = df_analisis['sentimen'].value_counts(dropna=False)
    print(sentiment_counts)

    # Visualisasi distribusi sentimen
    plt.figure(figsize=(8, 6))
    sns.barplot(x=sentiment_counts.index, y=sentiment_counts.values, palette='coolwarm')
    plt.title('Distribusi Sentimen')
    plt.xlabel('Sentimen')
    plt.ylabel('Jumlah Tweet')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    print("\nContoh data dengan sentimen:")
    display(df_analisis[['text_stemmed', 'sentimen', 'skor_sentimen']].head())
else:
    print("Kolom 'sentimen' dan 'skor_sentimen' tidak dibuat karena model gagal dimuat.")

print("\nAnalisis Sentimen Selesai.")
```

Hasil klasifikasi sentimen menggunakan model pre-trained w11wo/indonesian-roberta-base-sentiment-classifier mengungkap gambaran yang tidak mengejutkan namun tetap signifikan: sentimen **negatif mendominasi** percakapan publik tentang pelemahan rupiah. ### Topic Modeling

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
from nltk.corpus import stopwords

# Ensure stopwords are downloaded
try:
    nltk.data.find('corpora/stopwords')
except nltk.downloader.DownloadError:
    nltk.download('stopwords', quiet=True)

english_stopwords = set(stopwords.words('english'))
indonesian_stopwords = set(stopwords.words('indonesian'))
all_stopwords = english_stopwords.union(indonesian_stopwords)

sentiments = df_analisis['sentimen'].unique()

for sentiment_label in sentiments:
    if pd.notna(sentiment_label): # Skip NaN sentiments if any
        print(f"\n--- Topic Modeling for {sentiment_label.capitalize()} Sentiment ---")

        # Filter DataFrame for the current sentiment and drop NaNs from text_stemmed
        sentiment_df = df_analisis[df_analisis['sentimen'] == sentiment_label]
        corpus_sentiment = sentiment_df['text_stemmed'].dropna().tolist()

        if len(corpus_sentiment) < 2: # LDA needs at least 2 documents
            print(f"Tidak cukup data teks bersih untuk sentimen: {sentiment_label} ({len(corpus_sentiment)} dokumen). Melewati.")
            continue

        # Use CountVectorizer to create Document-Term Matrix (DTM)
        vectorizer = CountVectorizer(
            max_df=0.85, # ignore words that appear in 85% or more of documents
            min_df=2,    # ignore words that appear in less than 2 documents
            stop_words=list(all_stopwords)
        )

        try:
            dtm = vectorizer.fit_transform(corpus_sentiment)
        except ValueError as e:
            print(f"Tidak dapat membuat DTM untuk sentimen {sentiment_label}: {e}. Mungkin terlalu sedikit kata unik setelah stopword removal.")
            continue

        # Initialize and apply Latent Dirichlet Allocation (LDA)
        # Number of components (topics) can be adjusted. Start with 5 topics.
        n_components = min(5, dtm.shape[0] - 1) # Ensure n_components is less than number of documents
        if n_components < 2:
            print(f"Tidak cukup dokumen atau kata unik untuk membuat {n_components} topik untuk sentimen {sentiment_label}. Melewati.")
            continue

        lda = LatentDirichletAllocation(n_components=n_components, random_state=42)
        lda.fit(dtm)

        # Display top words for each topic
        feature_names = vectorizer.get_feature_names_out()
        for idx, topic in enumerate(lda.components_):
            top_words_idx = topic.argsort()[:-10-1:-1] # Get indices of top 10 words
            top_words = [feature_names[i] for i in top_words_idx]
            print(f"Topik {idx+1} ({sentiment_label.capitalize()}): {', '.join(top_words)}")

print("\nAnalisis Sentimen menggunakan Topic Modeling (LDA) Selesai.")
```

### Emotion Analysis

```python
from transformers import pipeline

emotion_model_loaded = False
try:
    # Load the specified emotion analysis model
    emotion_pipeline = pipeline(
        'text-classification',
        model='Aardiiiiy/NusaBERT-base-Indonesian-Plutchik-emotion-analysis',
        tokenizer='Aardiiiiy/NusaBERT-base-Indonesian-Plutchik-emotion-analysis',
        device=0 if torch.cuda.is_available() else -1 # Use GPU if available
    )
    emotion_model_loaded = True
    print("Model emosi berhasil dimuat: Aardiiiiy/NusaBERT-base-Indonesian-Plutchik-emotion-analysis")
except Exception as e:
    print(f"Gagal memuat model emosi: {e}")
    print("Analisis emosi akan dilewati karena masalah otorisasi/akses model atau model tidak ditemukan.")

if emotion_model_loaded:
    # Define a function to get emotion, handling potential NaN values and truncation
    def get_emotion(text):
        if pd.notna(text):
            try:
                # Apply emotion analysis, truncating long texts to prevent errors
                result = emotion_pipeline(text, truncation=True, max_length=512)[0]
                return result['label'], result['score']
            except Exception as e:
                print(f"Error during emotion prediction for text: {text[:50]}... Error: {e}")
                return None, None
        return None, None

    # Apply the function to 'text_stemmed' column and create new 'emosi' and 'skor_emosi' columns
    # Using .loc to avoid SettingWithCopyWarning if df_analisis is a slice
    df_analisis[['emosi', 'skor_emosi']] = df_analisis['text_clean'].apply(lambda x: pd.Series(get_emotion(x)))

    # Tampilkan distribusi emosi
    print("\nDistribusi Emosi:")
    emotion_counts = df_analisis['emosi'].value_counts(dropna=False)
    print(emotion_counts)

    # Visualisasi distribusi emosi
    plt.figure(figsize=(10, 7))
    sns.barplot(x=emotion_counts.index, y=emotion_counts.values, palette='viridis')
    plt.title('Distribusi Emosi')
    plt.xlabel('Emosi')
    plt.ylabel('Jumlah Tweet')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    print("\nContoh data dengan emosi:")
    display(df_analisis[['text_stemmed', 'emosi', 'skor_emosi']].head())
else:
    print("Kolom 'emosi' dan 'skor_emosi' tidak dibuat karena model gagal dimuat.")

print("\nAnalisis Emosi Selesai.")
```

Hasil analisis emosi menggunakan model Aardiiiiy/NusaBERT-base-Indonesian-Plutchik-emotion-analysis menghasilkan distribusi yang **sangat mengejutkan dan berlawanan dengan asumsi awal.** Dari total 24.910 tweet, distribusi emosi secara berurutan adalah sebagai berikut: **Trust (5.100)**, **Fear (5.050)**, **Anticipation (4.350)**, **Sadness (4.300)**, **Anger (2.300)**, **Surprise (1.950)**, **Disgust (1.150)**, dan **Joy (600)**.

**Untuk Peneliti Selanjutnya:**
Riset ini dapat dikembangkan ke tiga arah yang paling menjanjikan. Pertama, analisis cross-platform yang membandingkan sentimen di X, Instagram, dan TikTok untuk melihat apakah karakter emosional yang berbeda muncul di platform yang berbeda. Kedua, analisis time-series yang menghubungkan lonjakan volume tweet dan pergeseran distribusi emosi harian dengan pergerakan kurs aktual — untuk menguji apakah sentimen digital berfungsi sebagai leading indicator atau lagging indicator pergerakan rupiah. Ketiga, riset lanjutan yang secara spesifik membedah klaster narasi fiskal-politik (MBG-Prabowo) menggunakan metode analisis wacana (discourse analysis) untuk memahami lebih dalam bagaimana publik mengkonstruksi kausalitas antara kebijakan fiskal dan nilai tukar.
